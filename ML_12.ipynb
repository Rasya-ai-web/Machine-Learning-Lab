# Model Evaluation Using K-Fold Cross Validation

```python
import pandas as pd

from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
```

```python
df = pd.read_csv("iris_300.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
X = df[
    [
        "SepalLengthCm",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm"
    ]
]

y = df["Species"]
```

```python
model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="linear"))
])
```

```python
kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scores = cross_val_score(
    model,
    X,
    y,
    cv=kfold,
    scoring="accuracy"
)
```

```python
print("\nAccuracy for each fold:")

for i, score in enumerate(scores, start=1):
    print("Fold", i, ":", round(score, 4))
```

```python
print("\nAccuracy for each fold:")

for i, score in enumerate(scores, start=1):
    print("Fold", i, ":", round(score, 4))
```

```python
average_score = scores.mean()
```

```python
scores = cross_val_score(model, X, y, cv=5)
print("Cross-validation scores:", scores)
print("Mean Accuracy:", scores.mean())
```

```python
print("\n---------------------------------------")
print("K-FOLD CROSS VALIDATION RESULTS")
print("---------------------------------------")

print("Number of folds:", 5)

print("Average Accuracy:",
      round(average_score, 4))

print("Average Accuracy Percentage:",
      round(average_score * 100, 2), "%")

print("Standard Deviation:",
      round(scores.std(), 4))

print("\nModel performance is stable if the fold scores are close to each other.")
```

```python
print("\nK-Fold Cross Validation completed successfully!")
```
