STUDENT PERFORMANCE - SVM CLASSIFICATION

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

DATA LOADING

```python
df = pd.read_csv("student_performance_updated_1000.csv")
print("Dataset:")
print(df.head())
print("\nDataset Shape:", df.shape)
```

CREATE GRADE CATEGORY

```python
df["GradeCategory"] = pd.cut( df["FinalGrade"], bins=[-float("inf"), 50, 75, float("inf")], labels=["Low", "Medium", "High"] )
print("\nGrade Category Distribution:")
print(df["GradeCategory"].value_counts())
```

SELECT FEATURES AND TARGET

```python
X = df[ [ "AttendanceRate", "StudyHoursPerWeek", "PreviousGrade", "ExtracurricularActivities" ] ]
y = df["GradeCategory"]
```

REMOVE ROWS WHERE TARGET IS MISSING

```python
valid_target = y.notna()
X = X.loc[valid_target].copy()
y = y.loc[valid_target].copy()
```

TRAIN-TEST SPLIT

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=42, stratify=y )
print("\nTraining Data Shape:", X_train.shape)
print("Testing Data Shape :", X_test.shape)
```

IMPUTATION + SCALING + SVM

```python
svm_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler()),("svm", SVC( kernel="linear", random_state=42 )) ])
```

TRAIN MODEL

```python
svm_pipeline.fit(X_train, y_train)
print("\nSVM model trained successfully.")
```

PREDICTION

```python
y_pred = svm_pipeline.predict(X_test)
```

MODEL EVALUATION

```python
accuracy = accuracy_score(y_test, y_pred)
print("\nActual Values:")
print(y_test.values)
print("\nPredicted Values:")
print(y_pred)
print("\nAccuracy:", accuracy)
```

CLASSIFICATION REPORT

```python
print("\nClassification Report:\n")
print( classification_report( y_test, y_pred, zero_division=0 ) )
```

CHECK PREDICTED CLASSES

```python
print("\nActual Class Distribution:")
print(y_test.value_counts())
print("\nPredicted Class Distribution:")
print(pd.Series(y_pred).value_counts())
```
