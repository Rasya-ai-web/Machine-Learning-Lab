```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("/content/student_performance_updated_1000.csv")
```

```python
df.head(5)
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df = df.drop(["StudentID", "Name"], axis=1, errors="ignore")

df.head(5)
```

```python
df = pd.get_dummies(df, drop_first=True)

df.head(5)
```

```python
# Check and convert FinalGrade to numeric
df["FinalGrade"] = pd.to_numeric(df["FinalGrade"], errors="coerce")

# Remove rows where FinalGrade is missing
df = df.dropna(subset=["FinalGrade"]).copy()

# Convert FinalGrade into 3 classes
# 0 = Low
# 1 = Medium
# 2 = High

df["GradeClass"] = pd.cut(
    df["FinalGrade"],
    bins=[-float("inf"), 49, 74, float("inf")],
    labels=[0, 1, 2]
)

# Convert to integer
df["GradeClass"] = df["GradeClass"].astype(int)

# Display class counts
print("Grade Class Distribution:")
print(df["GradeClass"].value_counts().sort_index())
```

```python
X = df.drop(["FinalGrade", "GradeClass"], axis=1)

y = df["GradeClass"]

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
X = X.apply(pd.to_numeric, errors="coerce")

X = X.replace([float("inf"), float("-inf")], pd.NA)

X = X.fillna(X.median())

X = X.fillna(0)

print("Total Missing Values:", X.isnull().sum().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

print("Scaling completed successfully!")
```

```python
model = SVC(
    kernel="rbf",
    C=10,
    gamma="scale"
)

model
```

```python
model.fit(X_train, y_train)

print("SVM Model Training Completed!")
```

```python
y_pred = model.predict(X_test)

print("Prediction completed!")
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
print("Classification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1],
        target_names=[
            "Low Grade",
            "Medium Grade"
        ],
        zero_division=0
    )
)
```

```python
print("=" * 40)
print("       SVM CLASSIFICATION RESULT")
print("=" * 40)

print(f"Accuracy: {accuracy * 100:.2f}%")

if accuracy >= 0.90:
    print("Result: Accuracy is above 90%")
else:
    print("Result: Accuracy is below 90%")

print("=" * 40)
```
